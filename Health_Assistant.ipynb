```python
!pip install google-generativeai
import os
import google.generativeai as genai

#set APi Key
GOOGLE_API_KEY ="xxxxQ" # Please replace with your actual API key
genai.configure(api_key=(GOOGLE_API_KEY))

#configure model
# The error suggests 'gemini-pro' might not be available or supported for 'generateContent' in this environment.
# Please check the output above for available models that support 'generateContent',
# then replace 'gemini-pro' with a suitable model name.
model = genai.GenerativeModel('gemini-3.1-flash-lite-preview') # Corrected model name format

chat=model.start_chat()

#take symptom input
symptom=input("Describe your symptom: ")

#getting response
response=chat.send_message(f"The patient says {symptom}.Is this serious condition? what is your advice?")
print("AI Advice")
print(response.text)
```
