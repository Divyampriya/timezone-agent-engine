```python
import google.generativeai as genai
from google.colab import userdata
genai.configure(
    api_key=userdata.get('GOOGLE_API_KEY')
)

# List available models to find a suitable one
# for m in genai.list_models():
#   if 'generateContent' in m.supported_generation_methods:
#     print(m.name)

# Changing to 'gemini-pro' as it's a generally available and supported model
model = genai.GenerativeModel('gemini-3.1-flash-lite')
def simple_agent(prompt):
  # Define a local model for this function to use the specified preview version
  local_model = genai.GenerativeModel('gemini-3.1-flash-lite-preview')
  # Generate content and assign the result to the 'response' variable
  response = local_model.generate_content(prompt)
  return response.text
question ="explain difference between hardware and firmware"
answer = simple_agent(question)
print(answer)
```
